Q1. Creating a DataFrame
Create a Pandas DataFrame using a Python dictionary with the following columns:
Product
Region
Sales
Quantity

```python
import pandas as pd

# Create a dictionary
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Region': ['North', 'South', 'East', 'West'],
    'Sales': [50000, 12000, 15000, 30000],
    'Quantity': [10, 50, 30, 15]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print(df)
```

Q4. Indexing using .loc  

Using .loc, perform the following:
Display the row with index 10.
Display the Sales value of the row with index 20

```python
import pandas as pd

# Display the row with index 10
print("Row with index 10:")
print(df.loc[10])

# Display the Sales value of the row with index 20
print("\nSales value at index 20:")
print(df.loc[20, 'Sales'])
```

Q7. Filtering Data
Filter the dataset to display:
Rows where Region = "East".
Rows where Sales > 1500.

```python
import pandas as pd

# Rows where Region = "East"
print("Rows where Region = 'East':")
print(df[df['Region'] == 'East'])

# Rows where Sales > 1500
print("\nRows where Sales > 1500:")
print(df[df['Sales'] > 1500])
```

Q8. Multiple Condition Filtering
Display rows where:
Region is West
Sales is greater than 1200

```python
import pandas as pd

# Filter rows where Region is West and Sales > 1200
filtered_data = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]

# Display the result
print(filtered_data)
```

Q9. Descriptive Statistics
Generate summary statistics using:
describe()
From the result identify:
Mean sales value
Maximum quantity sold
Minimum sales value

```python
import pandas as pd

# Generate summary statistics
summary = df.describe()

print(summary)

# Extract specific values
mean_sales = summary.loc['mean', 'Sales']
max_quantity = summary.loc['max', 'Quantity']
min_sales = summary.loc['min', 'Sales']

print("\nMean Sales Value:", mean_sales)
print("Maximum Quantity Sold:", max_quantity)
print("Minimum Sales Value:", min_sales)
```

Q10. Aggregation Analysis
Using the Sales column, calculate:
Total sales
Average sales
Maximum sales
Minimum sales

```python
import pandas as pd

# Calculate aggregations on Sales
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
maximum_sales = df['Sales'].max()
minimum_sales = df['Sales'].min()

# Display results
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales:", maximum_sales)
print("Minimum Sales:", minimum_sales)
```
