### **Q1. What is the average, median, and standard deviation of house prices?**

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv("House Price India.csv")

# Calculate statistics for the Price column
average_price = df["Price"].mean()
median_price = df["Price"].median()
std_price = df["Price"].std()

print("Average House Price:", average_price)
print("Median House Price:", median_price)
print("Standard Deviation of House Price:", std_price)
```

Interpretation

Average (Mean) Price = 538,932.22

*  This represents the average cost of houses in the dataset.
*  On average, a house costs approximately 538,932.

Median Price = 450,000

*  Half of the houses have prices below 450,000 and half above.
*  Since the median is lower than the mean, the dataset likely contains some high-priced houses that increase the average.

Standard Deviation = 367,532.38

*  This measures how much house prices vary from the average.
*  A large standard deviation indicates that house prices are widely spread out, ranging from relatively inexpensive to very expensive properties.

Conclusion

The housing market in this dataset shows considerable price variation. The mean being higher than the median suggests a right-skewed distribution, meaning a small number of expensive houses are pulling the average upward.





### **Q2. Which number of bedrooms is most common?**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")

# Find the most common number of bedrooms
most_common_bedrooms = df["number of bedrooms"].mode()[0]

print("Most Common Number of Bedrooms:", most_common_bedrooms)

# Frequency of each bedroom count
bedroom_counts = df["number of bedrooms"].value_counts()
print(bedroom_counts)
```

Interpretation

*  The mode represents the bedroom count that appears most frequently in the dataset.
*  This indicates the type of house that is most commonly available in the housing market.

*  The frequency table helps compare the popularity of different bedroom configurations.





### **Q3. Check if the price data is skewed.**

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")
```

```python
print(df.columns)
```

```python
skewness = df['Price'].skew()

print("Skewness:", skewness)

if skewness > 0:
    print("The Price data is positively (right) skewed.")
elif skewness < 0:
    print("The Price data is negatively (left) skewed.")
else:
    print("The Price data is approximately symmetric.")
```

Interpretation



*   The skewness value is 4.2693, which is much greater than 0
*   The mean price (538,932) is higher than the median price (450,000).
*   This indicates that the distribution of house prices is strongly positively (right) skewed.
*   A small number of very expensive houses are pulling the average price upward.

### **Q5.  What is the relationship between living area and price? (Use : Area without Basement).**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")
```

```python
# Check column names first
print(df.columns)

# Correlation between Area without Basement and Price
correlation = df['Area of the house(excluding basement)'].corr(df['Price'])

print("Correlation:", correlation)
```

Interpretation



*   Correlation close to +1 → Strong positive relationship
*   Correlation between 0.5 and 0.8 → Moderate positive relationship


*   Correlation close to 0 → Weak or no relationship
*   Correlation close to -1 → Strong negative relationship





### **Q6. Identify any anomalies where houses have high prices but low areas.**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")
```

```python
# Calculate average price and average area
avg_price = df['Price'].mean()
avg_area = df['Area of the house(excluding basement)'].mean()

# Find anomalies: High Price but Low Area
anomalies = df[
    (df['Price'] > avg_price) &
    (df['Area of the house(excluding basement)'] < avg_area)
]

print("Number of anomalies:", len(anomalies))
print(anomalies[['Price', 'Area of the house(excluding basement)']])
```

Interpretation



*   These houses are priced above the average house price.
*   Their living area is below the average living area.
*   Such properties may be expensive due to factors other than size, such as:


*   Prime location
*   Luxury features
*   Recent renovations
*   Better neighborhood amenities
*   Waterfront or premium views













### **Q7. Compare average price based on number of floors and houses with or without waterfront.**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")
```

```python
avg_price_by_floors = df.groupby('number of floors')['Price'].mean()

print("Average Price by number of floors:")
print(avg_price_by_floors)
```

```python
avg_price_by_waterfront = df.groupby('waterfront present')['Price'].mean()

print("\nAverage Price by waterfront present:")
print(avg_price_by_waterfront)
```

```python
comparison = df.groupby(
    ['number of floors', 'waterfront present']
)['Price'].mean().reset_index()

print(comparison)
```

Interpretation

The average house price was compared across different numbers of floors and waterfront status. Houses with more floors generally tend to have higher average prices due to increased living space. Houses with waterfront access typically have significantly higher average prices than those without waterfront access, indicating that location and scenic value strongly influence property prices.

If the maximum price is much larger than the minimum price, it indicates that the dataset contains a wide range of properties, from affordable houses to luxury homes. This large gap often contributes to a positively skewed price distribution, where a few very expensive houses increase the overall average price.

### **Q9. Which location (zipcode/area) has the highest average price?**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("House Price India.csv")
```

```python
# Average price by zipcode
avg_price_zipcode = df.groupby('Postal Code')['Price'].mean()

# Zipcode with highest average price
highest_price_zipcode = avg_price_zipcode.idxmax()
highest_avg_price = avg_price_zipcode.max()

print("Postal Code with Highest Average Price:", highest_price_zipcode)
print("Highest Average Price:", highest_avg_price)
```

Interpretation

The zipcode/area with the highest average price represents the most expensive location in the dataset. Houses in this area are likely valued higher due to factors such as premium location, better amenities, accessibility, neighborhood quality, or demand.
