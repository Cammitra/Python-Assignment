**Q2 Sales Filtering Analysis
Filter the dataset to display:
Orders where Sales > 3500
Orders where Region = "East" and Sales > 2500
Explain what insights you observe.**

```python
# Filter orders with Sales greater than 3500
sales_above_3500 = df[df['Sales'] > 3500]

print(sales_above_3500)
```

```python
# Filter orders from East region with Sales greater than 2500
east_high_sales = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]

print(east_high_sales)
```

**Q3 Customer Purchase Analysis
Identify the top 5 customers based on total sales.
Hint: Use GroupBy with aggregation.**

```python
# Group by Customer and calculate total sales
top_customers = df.groupby('Customer')['Sales'].sum()

# Sort in descending order and get top 5 customers
top_5_customers = top_customers.sort_values(ascending=False).head(5)

print(top_5_customers)
```

**Q4 Regional Sales Performance
Calculate the total sales generated by each region.
Then identify:
Which region generated the highest revenue?**

```python
# Total sales generated by each region
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales)
```

```python
# Region with highest revenue
highest_region = region_sales.idxmax()
highest_sales = region_sales.max()

print("Highest Revenue Region:", highest_region)
print("Total Revenue:", highest_sales)
```

**Q5 Product Performance Analysis
Calculate the total quantity sold for each product.
Which product appears to be the most popular?**

```python
# Calculate total quantity sold for each product
product_quantity = df.groupby('Product')['Quantity'].sum()

print(product_quantity)
```

```python
# Product with the highest quantity sold
most_popular_product = product_quantity.idxmax()
highest_quantity = product_quantity.max()

print("Most Popular Product:", most_popular_product)
print("Total Quantity Sold:", highest_quantity)
```

**Q6 Multi-Level Grouping
Group the dataset by: Region and Category
Calculate the total sales for each combination.
What insights can you derive from this analysis?**

```python
# Group by Region and Category
region_category_sales = df.groupby(['Region', 'Category'])['Sales'].sum()

print(region_category_sales)
```

```python
# Find the Highest Sales Combination

highest_combo = (
    df.groupby(['Region', 'Category'])['Sales']
      .sum()
      .sort_values(ascending=False)
)

print(highest_combo.head(1))
```

**Insights**

-> Electronics vs Accessories

If Electronics sales are higher across most regions, it indicates that electronic products contribute the majority of revenue.
Electronics may be the primary revenue driver for the business.

-> Strongest Region-Category Combination

The combination with the highest sales (for example, East–Electronics) represents the most profitable market segment.
This segment may deserve additional marketing and inventory investment.

-> Regional Preferences

Some regions may purchase more Accessories while others focus on Electronics.
Understanding these preferences can help tailor promotions and stock levels by region.

-> Business Strategy

High-performing region-category combinations can be targeted for growth.
Low-performing combinations may require pricing changes, promotions, or product assortment adjustments.


**Q7 Pivot Table Creation
Create a pivot table showing:
Rows → Region
Columns → Product
Values → Total Sales
Which region sells the most laptops?**

```python
# Create pivot table

pivot_sales = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot_sales)
```

```python
# Which Region Sells the Most Laptops

laptop_sales = pivot_sales['Laptop']

print(laptop_sales)
print("Region with highest laptop sales:", laptop_sales.idxmax())
print("Total laptop sales:", laptop_sales.max())
```

**Q8 Profitability Analysis
Calculate the average profit for each product category.
Which category is most profitable?**

```python
# Calculate average profit by category
category_profit = df.groupby('Category')['Profit'].mean()

print(category_profit)
```

```python
# Most profitable category
most_profitable = category_profit.idxmax()
highest_avg_profit = category_profit.max()

print("Most Profitable Category:", most_profitable)
print("Average Profit:", highest_avg_profit)
```

**Q9 Discount Impact Study
Analyze the relationship between discount and profit.
Tasks:
Calculate the average profit for each discount level
Identify whether higher discounts lead to lower profit**

```python
# Average profit by discount level
discount_profit = (
    df.groupby('Discount')['Profit']
      .mean()
      .reset_index()
      .sort_values(by='Discount')
)

print(discount_profit)
```

```python
# Correlation between Discount and Profit
correlation = df['Discount'].corr(df['Profit'])

print("Correlation:", correlation)
```

**Q10 End-to-End Data Wrangling Task
Perform the following workflow:
Filter the dataset where Sales > 3000
Group data by Region and Product
Calculate total sales
Reshape the result into a pivot table
Explain the business insights obtained from this analysis.**

```python
# Filter the Dataset (Sales > 3000)

high_sales = df[df['Sales'] > 3000]

print(high_sales.head())
```

```python
# Group Data by Region and Product

grouped_sales = (
    high_sales.groupby(['Region', 'Product'])['Sales']
    .sum()
)

print(grouped_sales)
```

```python
# Calculate total sales
total_sales = df['Sales'].sum()

print("Total Sales:", total_sales)
```

```python
# Reshape into a Pivot Table

pivot_sales = pd.pivot_table(
    high_sales,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot_sales)
```

**Business Insights**

-> Focus on High-Value Transactions

By filtering Sales > 3000, the analysis focuses only on large orders that contribute significantly to revenue.


-> Identify Top-Selling Products by Region

The pivot table shows which products generate the most revenue in each region.

East may generate the highest Laptop sales.
South may perform strongly in Mouse and Tablet sales.
West may have strong Keyboard sales.



-> Regional Product Preferences

Different regions often prefer different products:

These patterns help with inventory planning and targeted marketing.
